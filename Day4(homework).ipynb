```python
#1 Positive or negative until user enters 0
# Keep taking a number form the user. Print whether it's **positive** or **negative**.
# stop when the user enters **0**.
```

```python
num = int(input("Enter a number (0 for stop) :- "))
while num != 0:
  if num >=0:
    print("positive")
  else:
    print("negative")
  num = int(input("Enter a number (0 for stop) :- "))
```

```python
# Even or Odd checker (loop continues)** Ask for a number repeatedly.
# Print **"Even"** or **"Odd"** each time. Stop only when user enters **-1**.
```

```python
num = int(input("Enter a number (-1 for stop) :- "))
while num != -1:
  if num >= -1:
    print("Even")
  else:
    print("Odd")
  num = int(input("Enter a number (-1 for stop) :- "))
```

```python
#5 Print numbers up to user limit. Tke input 'n'.
# Use a 'while' loop to print all numbers from **1 to n**.
# If 'n<=0', print "Invalid number".
```

```python
n = int(input("Enter a number :- "))
if n <= 0:
  print("Invalid number")
else:
  i = 1
  while i <= n:
    print(i)
    i += 1
```
