```python
import numpy as np
import pandas as pd
```

```python
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
```

```python
titanic.head(1)
```

```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
# show charts inside notebook
```

```python
# HISTOGRAM - Distribution of age
```

```python
plt.figure(figsize=(8,4))
plt.hist(titanic["age"].dropna(), bins=30, color='skyblue', edgecolor='black')
plt.title("Age Distribution Historgam", fontsize=20, fontweight='bold')
plt.xlabel("Age")
plt.ylabel("Frequency or Count")
plt.legend(['Age'])
plt.show()
```

```python
# BAR CHART
```

```python
X1= titanic['sex'].value_counts()
```

```python
X1.values
```

```python
X1.index
```

```python
plt.figure(figsize=(6,4))
X1.plot(kind='bar', color=['pink', 'red'])
plt.xticks(rotation=45)
```

```python
# Pie Chart
```

```python
X2 = titanic['survived'].value_counts()
```

```python
X2
```

```python
plt.figure(figsize=(5,5))
plt.pie(X2, labels=['Not Survived', 'Survived'],
        autopct='%1.2f%%', colors=['red', 'green'])
plt.legend()
```
