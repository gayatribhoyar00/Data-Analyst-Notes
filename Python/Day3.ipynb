```python
# FOR LOOP - when we have a range - that mean we have start also and end also
```

```python
fruits = ['apple', 'banana', 'cherry', 'kiwi']
```

```python
# Iteration mean one by one accessing the element
# in - keyword for iteration
```

```python
for i in fruits:
  print(i)
```

```python
for i in range(1, 11):
  print(i)
```

```python
# indentation - we need extra space in next line after :
```

```python
total = 0
for i in range(1, 101):
  total = total + i
```

```python
total
```

```python
num = 5
for i in range(1, 11):
  print(f'{num} * {i} = {num * i}')
```

```python
# nested loop (loop between loop)
```

```python
for num in range(5, 8):
  for i in range(1, 11):
    print(f'{num} * {i} = {num * i}')
  print("-----------------")
```
