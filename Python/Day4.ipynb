```python
num = int(input("Enter a number (0 for stop) :- "))
while num != 0:
  if num >=0:
    print("positive")
  else:
    print("negative")
  num = int(input("Enter a number (0 for stop) :- "))
```

```python
num = int(input("Enter a number (0 for stop) :- "))
while num != 0:
  if num % 2 == 0:
    print("even no")
  else:
    print("odd no")
  num = int(input("Enter a number (0 for stop) :- "))
```
